`Motivation to Derivatives`

- A Derivate is instantaneous rate of change of function

- Formula for calculating slope  = rise / run

- Derivative = dy / dx

- The tangent to a curve at a point is the line that best approximates the curve at that point.

- This means that the tangent line has the same slope as the curve at that point, and it passes through the curve at that point.

`Slopes,maxima & minima`

* Slope = Velocity
* Maximum or Minimum of a function occurs at a place where velocity = 0 or tangent line is horizontal

**Slope**

✔️ The slope of a line at a point is the rate of change of the function at that point.
✔️ The slope of a tangent line is the instantaneous rate of change of the function at a point.

**Maxima and minima**

✔️ A maximum is a point where the function is greater than or equal to all of its neighbors.
✔️ A minimum is a point where the function is less than or equal to all of its neighbors.
✔️ A local maximum/minimum is a maximum/minimum relative to its nearby points.
✔️ A global maximum/minimum is a maximum/minimum relative to the entire function.
