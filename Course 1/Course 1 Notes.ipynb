`System of Sentences :`

A system of sentences in linear algebra is a collection of one or more linear equations involving the same variables.

- `Non-Singular` - When the sentence contains many different information ( also known as complete)

- `Singular` - the system contains less info than Non-singular one (also known as Redundant,Contradictory)

`System of Equations` 

A system of equations in linear algebra is a collection of one or more linear equations involving the same variables. A linear equation is an equation of the form ax+by=c, where a, b, and c are constants and x and y are variables.

If we get Unique solution to a system its `Non Singular` ,For Infinte and NO solutions system is `Singular`

`System of equations as lines`

In linear algebra, a system of two linear equations in two variables can be represented by a pair of lines in the plane. The solution to the system of equations is the point where the two lines intersect.

`Linear Dependence and Independence`

- A set of vectors is said to be linearly dependent if one of the vectors can be expressed as a linear combination of the other vectors in the set

- A set of vectors is said to be linearly independent if none of the vectors in the set can be expressed as a linear combination of the other vectors in the set. 

`The Determinant`

Deteminant returns `0` for Singular matrix & `number other than 0` for non singular `Determinant = ad-bc`
